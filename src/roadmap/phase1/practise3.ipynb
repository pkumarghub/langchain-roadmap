```python
from langchain_core.prompts import ChatPromptTemplate

# The following will only print the prompt template which has been created for AI model.
template = ChatPromptTemplate(
    [
        ("system", "You are a helpful AI bot. user will have a prompt and you have to tell about it."),
        ("human", "What is {user_input}"),
    ]
)

prompt_response = template.invoke("html")

print(prompt_response)
```

```python
from langchain_groq import ChatGroq

template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful AI bot. User will give you a topic and you have to explain it in 20 words"),
        ("human", "What is {user_input}"),
    ]
)

final_prompt = template.format_messages(user_input= "html")

# Create the chat model
llm = ChatGroq(
    model="llama-3.3-70b-versatile",  # or any Groq-supported model
    temperature=0
)

prompt_response = llm.invoke(final_prompt);

print(prompt_response.content)
```
