Exercise 1: Say Hello to the LLM ⭐
Goal:- Learn how to use a LangChain Chat Model.

Problem:- Write a program that asks the LLM:

"What is Python?"

Print the response.

Concepts
ChatOpenAI (or ChatAnthropic/Gemini)
Basic model invocation
Don't Use
PromptTemplate
LCEL
Parser

```python
from langchain_groq import ChatGroq

# Create the chat model
llm = ChatGroq(
    model="llama-3.3-70b-versatile",  # or any Groq-supported model
    temperature=0
)

# Ask the question
response = llm.invoke("What is Python?")

# Print the response
print(response.content)
```
