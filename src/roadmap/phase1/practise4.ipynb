Exercise 4: Multiple Variables ⭐⭐
Goal:- Use more than one variable.

Prompt
Explain {topic} to a {level} student.

Examples

topic = "Python"
level = "Beginner"
topic = "Docker"
level = "Expert"

Concepts
Prompt variables
Reusable prompts

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq
import os
from dotenv import load_dotenv

load_dotenv()

template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful AI bot. Answer the user."),
        ("human", "Explain {topic} to a {level} student."),
    ]
)

formated_message = template.format_messages(
    topic="computer science",
    level="4"
)

llm = ChatGroq(
    model=os.getenv("CHAT_MODEL_NAME"),
    temperature=0
)

prompt_response = llm.invoke(formated_message)

print(prompt_response.content)
```
