Your First LCEL Chain ⭐⭐⭐

Goal:- Write your first LangChain chain.

Instead of

prompt.invoke()
llm.invoke()
parser.invoke()

Build
prompt | llm | parser

Ask
Explain Git.

Concepts:-
LCEL
Runnable
StrOutputParser
JsonOutputParser

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser, JsonOutputParser
from pydantic import BaseModel, Field

from dotenv import load_dotenv
import os

load_dotenv()

llm = ChatGroq(
    model=os.getenv("CHAT_MODEL_NAME"),
    temperature=0
)


class JsonFormat(BaseModel):
    introduction: str = Field(
        description="Explain about the topic in 20 words")
    purpose: str = Field(description="How do we use it")
    benefits: int = Field(description="benefits of using git")


parser = StrOutputParser()
json_parser = JsonOutputParser(pydantic_object=JsonFormat)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI bot. User will give you a question and you have to explain it in 20 words"),
    ("human", "Explain {topic}"),
])

json_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a helpful AI assistant. Return your answer in the following JSON format. {format_instructions} """
    ),
    (
        "human", "Explain {topic}."
    ),
]
).partial(
    format_instructions=json_parser.get_format_instructions()
)

chain = prompt | llm | parser
json_chain = json_prompt | llm | json_parser

response = chain.invoke({
    "topic": "git"
})

json_response = json_chain.invoke({
    "topic": "git"
})

print("----- Text Output -----")
print(response)

print("\n----- JSON Output -----")
print(json_response)
```
