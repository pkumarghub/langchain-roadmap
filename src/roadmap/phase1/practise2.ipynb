Exercise 2: Ask Different Questions ⭐
Goal:- Understand that the model is reusable.

Problem
Store these questions in a list:

questions = [
    "What is Python?",
    "What is Java?",
    "What is C++?"
]

Loop through the list and ask the LLM each question.

Print every response.

Concepts
Chat Model
Multiple invocations

```python
from langchain_groq import ChatGroq

# Create the chat model
llm = ChatGroq(
    model="llama-3.3-70b-versatile",  # or any Groq-supported model
    temperature=0
)

questions = [
    "What is Python? Answer in 15 words.",
    "What is Java? Answer in 15 words.",
    "What is C++? Answer in 15 words."
]

for question in questions:
    response = llm.invoke(question)
    print(response.content)

```
